# ResNet CIFAR-100 Training from Scratch

## 🎯 Assignment Objectives
- Train ResNet model from scratch on CIFAR-100
- Target: **73% top-1 accuracy**
- Training: ~100 epochs
- No pre-trained models allowed
- Deploy to HuggingFace Spaces

## 📋 CIFAR-100 Dataset Info
- **Classes**: 100 fine-grained classes
- **Training samples**: 50,000 (500 per class)
- **Test samples**: 10,000 (100 per class)
- **Image size**: 32×32×3
- **Superclasses**: 20 coarse classes
